# Сборка общей таблицы признаков

Объеденение сохранённых признаков.


## 1. Загрузка результатов


```python
import pandas as pd

features = pd.read_parquet('data/name_features.parquet')
number_features = pd.read_parquet('data/number_features.parquet')
brand_features = pd.read_parquet('data/brand_features.parquet')
weight_features = pd.read_parquet('data/weight_features.parquet')
article_feauters = pd.read_parquet('data/pairs_attrs.parquet')
```

## 2. Объединение признаков

Присоединяем числа, бренд и массу по двум идентификаторам пары.

```python
features = features.merge(
    number_features[['id1', 'id2', 'numbers_jaccard']],
    on=['id1', 'id2'],
    how='left',
    validate='one_to_one'
)

features = features.merge(
    brand_features[['id1', 'id2', 'brand_common']],
    on=['id1', 'id2'],
    how='left',
    validate='one_to_one'
)

features = features.merge(
    weight_features[['id1', 'id2', 'weight_rel_diff']],
    on=['id1', 'id2'],
    how='left',
    validate='one_to_one'
)

features = features.merge(
    article_feauters[['id1', 'id2', 'article_equal', 'partnumber_equal', 'brand_equal']],
    on=['id1', 'id2'],
    how='left',
    validate='one_to_one'
)
```

## 3. Проверка результата




```python
print(features.shape)
print(features.columns)
features.notna().sum()
```

## 4. Сохранение общей таблицы


```python
features.to_parquet('data/features.parquet', index=False)
```
