## Подготовка таблицы pairs


импорты

```python
import numpy as np
import pandas as pd

```

```python
matches_data = pd.read_parquet('data/matches.parquet')
items_human_data = pd.read_parquet('data/items_human.parquet')
```

```python
print(matches_data.head())
```

```python
print(items_human_data.head())
```

```python
print(items_human_data['id'].nunique())
print(len(items_human_data['id']))
```

```python
pairs = matches_data.merge(items_human_data, left_on='id1', right_on='id', how='left')
pairs = pairs.merge(items_human_data, left_on='id2', right_on='id', how='left', suffixes=('_1', '_2'))
```

```python
print(pairs.head())
```

```python
print(pairs['name_1'].isna().sum())
print(pairs['name_2'].isna().sum())
```

```python
print((pairs['id1'] == pairs['id_1']).sum() == len(pairs))
print((pairs['id2'] == pairs['id_2']).sum() == len(pairs))
```

```python
pairs = pairs.drop(columns=['id_1', 'id_2'])
```

```python
print((pairs['category_1'] == pairs['category_2']).sum())
```

```python
pairs = pairs.drop(columns=['category_2'])
```

```python
pairs = pairs.rename(columns={'category_1': 'category'})
print(pairs.columns)
```

```python
pairs = pairs[['id1', 'id2', 'target', 'category', 'name_1', 'name_2', 'attributes_1', 'attributes_2']]
```

```python
pairs.to_parquet('data/pairs.parquet')
```
